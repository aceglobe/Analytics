```python
import re   #module for regex

#Introduction to Python
#Data types in Python
x = 2      # datatype of x is integer
print(x)

y = 2.15    # datatype of y is float
print(y)

z = "Hello World, Python!"  # datatype of z is a string 
u = "1"; v = "12" 
print(int(v))

bool1 = True           #datatype of bool1 is boolean
print(bool1)
type(bool1)

#We are able to change the type of
#One datatype to another using a process
#called TYPE CASTING


print(int(y))    #changes the float y datatype to int

print(float(x))   #changes the int x datatype to float

print(str(x))   #type casting of int x to string 
type(str(x))

#Manipulating Strings
#There are different formats by which we can print strings
# New approach used in python version 3.6 and later is f-string

#f-string
print(f"{z}, \nthis is my first python code")

#Using str.format
print("This is {}".format(z))

#Using % 
print("This adds an int %d and a float %1.2f in my %s program" %(x,y,z)) 

#Raw string
#sometimes some strings have paths
#we use raw strings to preserve the format
path = "This is a path \n"   #Eg. this doesn't print the path we need
print(path)

print(r"This is a path \n")  #Raw string preserves the format

#String Manipulation
#We can change the case of strings using <name>.upper or <name>.lower()
z.upper()
z.lower()

#We can search for a string using find
z.find("Python")   #this prints the index of the string being searched 

#We can also split a string 
z.split()  #this uses whitespace to split the string

#we are also able to split string by specifying a separator and or maxsplit
z.split("Hello",2)

#RegEx (Regular Expression)
S1 = "The Body Guard is good!"
pattern = "Body"
result= re.search(pattern,S1)
#checking if the search found a match in the string S1

if result:
    print("match found")
    print("Matched string is:", result.group())
else: 
    print("No match found")

#Data Structures
#These help us to save data in python
#There are different types of data structures
#Lists # Tuples #Dictionaries 

#Tuple
#they are used in storing datapoints in python
#we are not able to modify the elements of tuples after defined/initialized
tup1 = (1,2,4,5)
print(tup1)

#We can also define tuples without the parathenses
tup2 = 1,2,3
print(tup2)


#List
#These are a different data structure that can be used in storing data in python
#We are able to modify the elements of lists after definition
list1 = [1,3,4,6]
print(list1)

#Dictionary
#These comprise key and value pairs 
dict1 = {"Age": [25,17.54], "Gender": ["Male", "Male", "Female"]}
print(dict1)
```
