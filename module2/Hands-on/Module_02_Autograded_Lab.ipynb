```python
# YOUR CODE HERE
avg_age = df['Age'].mean()
avg_income = df['Annual_Income'].mean()
avg_age, avg_income
```

## Question 3
**Group by Gender and compute churn rate**

```python
# YOUR CODE HERE
churn_rate = df.groupby('Gender')['Churn'].mean()
churn_rate
```

```python
assert churn_rate.shape[0] >= 2
```

## Question 4

```python
# YOUR CODE HERE
import matplotlib.pyplot as plt
churn_rate.plot(kind='bar')
plt.title('Churn Rate by Gender')
plt.show()
```
